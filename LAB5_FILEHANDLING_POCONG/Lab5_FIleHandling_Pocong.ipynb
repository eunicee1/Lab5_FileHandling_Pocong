```python
# Step 1: Import additional Libraries
from datetime import datetime
from pathlib import Path
import shutil

# Step 2: Define working directory
backup_dir = Path.home() / "Documents" / "PythonFiles"
backup_dir.mkdir(exist_ok=True)

USERNAME = "pocong"   # <-- your username

# Step 3: Function to write a file with backup
def write_with_backup(filename: str, content: str, mode="w"):

    file_path = backup_dir / filename

    # create backup if file exists
    if file_path.exists():
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

        # backup filename format with username
        backup_name = (
            f"{file_path.stem}_{USERNAME}_{timestamp}{file_path.suffix}"
        )

        backup_path = backup_dir / backup_name
        shutil.copy2(file_path, backup_path)

        print(f"Backup saved: {backup_path.name}")

    # write or append
    with open(file_path, mode, encoding="utf-8") as file:
        file.write(content + "\n")

    print(f"File saved: {file_path.name}")


# Step 4: Function to read file
def read_file(filename: str):
    file_path = backup_dir / filename
    with open(file_path, "r", encoding="utf-8") as file:
        return file.read()


# ✅ NEW: Function to list backups of a specific file
def list_backups(filename: str):
    print("\nBackups found:")
    stem = Path(filename).stem

    found = False
    for backup in backup_dir.glob(f"{stem}_{USERNAME}_*"):
        print("-", backup.name)
        found = True

    if not found:
        print("No backups available.")


# Step 5: Run demonstration
print("=== File Operations Demo ===")

# Create file
print("\n1. Creating new file:")
write_with_backup("demo.txt", "Initial content")

# Ask user overwrite or append
choice = input("\nOverwrite or Append? (w/a): ").lower()

if choice == "a":
    mode = "a"
else:
    mode = "w"

print("\n2. Updating file (with backup):")
write_with_backup("demo.txt", "Updated content", mode)

# Read file
print("\n3. Reading file:")
print(read_file("demo.txt"))

# List backups for demo.txt
print("\n4. Listing backups for demo.txt")
list_backups("demo.txt")
```

```python
# Step 1: Import required modules
from pathlib import Path
from datetime import datetime
import shutil

# Step 2: Define your custom working directory in Documents
backup_dir = Path.home() / "Documents" / "PythonFiles"
backup_dir.mkdir(exist_ok=True) # Create folder if it doesn't exist

# Step 3: Define the menu-driven file manager
def file_manager():
    file_name = input("Enter filename (e.g., notes.txt): ")
    file_path = backup_dir / file_name

    while True:
        print("\n--- MENU ---")
        print("1. Write to file")
        print("2. Append to file")
        print("3. Read file")
        print("4. Backup file")
        print("5. Exit")

        choice = input("Choose an option (1-5): ")

        if choice == "1":
            content = input("Enter content to write:\n")
            with open(file_path, "w", encoding="utf-8") as f:
                f.write(content)
            print("File written successfully.")

        elif choice == "2":
            more = input("Enter content to append:\n")
            with open(file_path, "a", encoding="utf-8") as f:
                f.write("\n" + more)
            print("Content appended.")

        elif choice == "3":
            if file_path.exists():
                with open(file_path, "r", encoding="utf-8") as f:
                    print("\nFile Content:\n", f.read())
            else:
                print("File not found.")

        elif choice == "4":
            if file_path.exists():
                timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
                backup_file = file_path.with_name(
                    f"{file_path.stem}_backup_{timestamp}{file_path.suffix}"
                )
                shutil.copy2(file_path, backup_file)
                print(f"Backup created: {backup_file.name}")
            else:
                print("Cannot backup. File does not exist.")

        elif choice == "5":
            print("Exiting the file manager.")
            break

        else:
            print("Invalid choice. Please try again.")

# Step 4: Run the menu-driven tool
if __name__ == "__main__":
    file_manager()
```

```python
from pathlib import Path
from datetime import datetime
import shutil

# 1. Change the working directory to Surname_Activity_5
# Replace 'Doe' with your actual surname
surname = "Doe" 
working_dir = Path.home() / "Documents" / f"{surname}_Activity_5"
working_dir.mkdir(exist_ok=True)

def file_manager():
    # 4 & 5. Test with different names and extensions (e.g., .csv, .md)
    file_name = input("Enter filename with extension (e.g., data.csv, notes.md): ")
    file_path = working_dir / file_name

    while True:
        print(f"\n--- {surname.upper()}'S FILE MANAGER ---")
        print("1. Write to file")
        print("2. Append to file")
        print("3. Read file")
        print("4. Backup file")
        print("5. List all backup files") # Added 6th option logic
        print("6. Exit")

        choice = input("Choose an option (1-6): ")

        if choice == "1":
            content = input("Enter content to write:\n")
            with open(file_path, "w", encoding="utf-8") as f:
                f.write(content)
            print("File written successfully.")

        elif choice == "2":
            more = input("Enter content to append:\n")
            with open(file_path, "a", encoding="utf-8") as f:
                f.write("\n" + more)
            print("Content appended.")

        elif choice == "3":
            if file_path.exists():
                with open(file_path, "r", encoding="utf-8") as f:
                    print("\nFile Content:\n", f.read())
            else:
                print("File not found.")

        elif choice == "4":
            if file_path.exists():
                timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
                # 3. Modify backup name to include surname
                backup_name = f"{file_path.stem}_{surname}_backup_{timestamp}{file_path.suffix}"
                backup_file = file_path.with_name(backup_name)
                
                shutil.copy2(file_path, backup_file)
                print(f"Backup created: {backup_file.name}")
            else:
                print("Cannot backup. File does not exist.")

        elif choice == "5":
            # 2. Add a 6th menu option to list all backup files
            print(f"\n--- Backup Files in {working_dir.name} ---")
            backups = list(working_dir.glob(f"*_{surname}_backup_*"))
            if backups:
                for b in backups:
                    print(f"- {b.name}")
            else:
                print("No backup files found.")

        elif choice == "6":
            print("Exiting the file manager.")
            break

        else:
            print("Invalid choice. Please try again.")

if __name__ == "__main__":
    file_manager()
```
