```python
import os

path = "PythonFiles"
files = [f for f in os.listdir(path) if f.endswith(".txt")]

data = []

for f in files:
    with open(os.path.join(path, f), "r", encoding="utf-8") as file:
        lines = [l.strip() for l in file if l.strip()]
        words = sum(len(l.split()) for l in lines)
        chars = sum(len(l.replace(" ", "")) for l in lines)

    wl = words / len(lines) if lines else 0
    cw = chars / words if words else 0
    data.append((f, len(lines), words, chars, wl, cw))

# sort by word density (words/line)
data.sort(key=lambda x: x[4], reverse=True)

print("Filename | Lines | Words | Chars | W/L | C/W")
for d in data:
    print(f"{d[0]} | {d[1]} | {d[2]} | {d[3]} | {d[4]:.2f} | {d[5]:.2f}")

print("\nMost dense:", data[0][0])
print("Least dense:", data[-1][0])
```

```python
import os, re
from collections import Counter

path = "PythonFiles"

# load stopwords (optional)
try:
    stop = set(open("stopwords.txt").read().split())
except:
    stop = set()

total = Counter()

for f in os.listdir(path):
    if f.endswith(".txt"):
        text = open(os.path.join(path, f), encoding="utf-8").read().lower()
        words = re.findall(r'\b\w+\b', text)
        words = [w for w in words if w not in stop]
        total.update(words)

print("Most frequent words:")
for w, c in total.most_common(10):
    print(w, c)
```

```python
import os, shutil
from datetime import datetime

student_id = "TUPM-25-0591"
student_name = "EUNICE G. POCONG"

src = "PythonFiles.txt"

try:
    if not os.path.isfile(src):
        print("File not found:", src)
    else:
        backup_dir = f"Documents/Activity_5_Exercises/backup_{student_id}"
        os.makedirs(backup_dir, exist_ok=True)

        ts = datetime.now().strftime("%Y%m%d_%H%M%S")
        base = os.path.splitext(os.path.basename(src))[0]
        dest = os.path.join(backup_dir, f"{base}_{student_id}_{ts}.txt")

        shutil.copy(src, dest)

        log = os.path.join(backup_dir, f"backup_log_{student_id}.txt")
        size = os.path.getsize(dest)

        with open(log, "a") as f:
            f.write(f"{student_id} | {student_name} | {base}.txt | {ts} | {size} bytes\n")

        print("Backup successful!")
        print("Saved as:", dest)

except Exception as e:
    print("Error:", e)
```
