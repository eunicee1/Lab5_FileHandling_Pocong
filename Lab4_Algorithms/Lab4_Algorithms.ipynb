```python
# Step 1: Implement bubble sort
def bubble_sort(arr):
    n = len(arr)
    swaps = 0   # counter for swaps

    for i in range(n):
        for j in range(0, n - i - 1):

            if arr[j] > arr[j + 1]:
                # swap values
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swaps += 1   # count swap

    return swaps


# Step 2: Test sorting
to_sort = [64, 34, 25, 12, 22, 11, 90]

swap_count = bubble_sort(to_sort)

print("Sorted list:", to_sort)
print("Number of swaps:", swap_count)
```

```python
def selection_sort_count(arr):
    comparisons = 0
    swaps = 0

    for i in range(len(arr)):
        min_idx = i

        for j in range(i+1, len(arr)):
            comparisons += 1

            if arr[j] < arr[min_idx]:
                min_idx = j

        if min_idx != i:
            arr[i], arr[min_idx] = arr[min_idx], arr[i]
            swaps += 1

    return comparisons, swaps


sample_data = [29, 10, 14, 37, 13]

comp, swp = selection_sort_count(sample_data)

print("Sorted list:", sample_data)
print("Comparisons:", comp)
print("Swaps:", swp)
```

```python
# Step 1: Implement insertion sort
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >=0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key

arr = [12, 11, 13, 5, 6]
insertion_sort(arr)
print("Sorted array:", arr)
        
```

```python
# E ADDITIONAL EXCERCISES
#1. Algorithm Complexity Profiler
import time

# -------------------------
# TUPM-25-0591
# -------------------------
student_id = "TUPM-25-0591"

# target = last two digits
target = int(student_id[-2:])

# main list (sorted for binary search)
main_list = list(range(1, 5000))

# small list with duplicate
small_list = [1,2,3,4,5,6,3,7,8]

# -------------------------
# Linear Search
# -------------------------
def linear_search(arr, key):
    for i in arr:
        if i == key:
            return True
    return False

# -------------------------
# Binary Search
# -------------------------
def binary_search(arr, key):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == key:
            return True
        elif arr[mid] < key:
            low = mid + 1
        else:
            high = mid - 1

    return False

# -------------------------
# Quadratic Duplicate Finder
# -------------------------
def find_duplicate(arr):
    for i in range(len(arr)):
        for j in range(i+1, len(arr)):
            if arr[i] == arr[j]:
                return arr[i]

# -------------------------
# Timing
# -------------------------
start = time.time()
linear_search(main_list, target)
linear_time = time.time() - start

start = time.time()
binary_search(main_list, target)
binary_time = time.time() - start

start = time.time()
find_duplicate(small_list)
quadratic_time = time.time() - start

# -------------------------
# OUTPUT TABLE
# -------------------------
print("\nAlgorithm Complexity Profiler")
print("------------------------------------")
print("Linear Search Time:", linear_time)
print("Binary Search Time:", binary_time)
print("Quadratic Algo Time:", quadratic_time)

print("\nAnalysis:")
print("Binary search scales better because it reduces the search space each step.")
```

```python
#Recursive vs. Iterative Strategy Analysis
import time

# -------------------------
# STUDENT ID
# -------------------------
student_id = "TUPM-25-0591"

n = int(student_id[-1]) + 15

# -------------------------
# Iterative Fibonacci
# -------------------------
def fibonacci_iterative(n):
    a, b = 0, 1

    for _ in range(n):
        a, b = b, a + b

    return a

# -------------------------
# Timing
# -------------------------
start = time.time()
result = fibonacci_iterative(n)
elapsed_time = time.time() - start

# -------------------------
# OUTPUT
# -------------------------
print("\nRecursive vs Iterative Analysis")
print("--------------------------------")
print("Problem Size (n):", n)
print("Fibonacci Result:", result)
print("Execution Time:", elapsed_time)

print("\nDiscussion:")
print("Iterative Fibonacci runs in linear time O(n).")
```

```python
#Sorting Algorithm Visual Comparator
import random
import time

# -------------------------
# STUDENT ID
# -------------------------
student_id = "20241234"

size = int(student_id[-3:]) + 50

data = random.sample(range(1, 10000), size)

# -------------------------
# Bubble Sort
# -------------------------
def bubble_sort(arr):
    comp = swap = 0
    start = time.time()

    n = len(arr)
    for i in range(n):
        for j in range(0, n-i-1):
            comp += 1
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swap += 1

    return comp, swap, time.time()-start

# -------------------------
# Selection Sort
# -------------------------
def selection_sort(arr):
    comp = swap = 0
    start = time.time()

    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            comp += 1
            if arr[j] < arr[min_idx]:
                min_idx = j

        arr[i], arr[min_idx] = arr[min_idx], arr[i]
        swap += 1

    return comp, swap, time.time()-start

# -------------------------
# Insertion Sort
# -------------------------
def insertion_sort(arr):
    comp = swap = 0
    start = time.time()

    for i in range(1, len(arr)):
        key = arr[i]
        j = i-1

        while j >= 0 and arr[j] > key:
            comp += 1
            arr[j+1] = arr[j]
            j -= 1
            swap += 1

        arr[j+1] = key

    return comp, swap, time.time()-start

# -------------------------
# RUN ALGORITHMS
# -------------------------
b = bubble_sort(data.copy())
s = selection_sort(data.copy())
i = insertion_sort(data.copy())

# -------------------------
# OUTPUT
# -------------------------
print("\nSorting Algorithm Comparison")
print("----------------------------------------------")
print("Algorithm | Comparisons | Swaps | Time")
print("Bubble    |", b)
print("Selection |", s)
print("Insertion |", i)

print("\nRecommendation:")
print("Insertion sort is usually faster for small datasets.")
```
